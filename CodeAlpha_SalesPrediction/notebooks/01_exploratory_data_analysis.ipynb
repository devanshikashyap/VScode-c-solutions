# Sales Prediction - Exploratory Data Analysis (EDA)

This notebook explores the **Advertising Dataset** to analyze how marketing budgets across **TV**, **Radio**, and **Newspaper** impact product sales.

### Objectives:
1. Ingest advertising dataset and verify data hygiene (missing values, types).
2. Inspect summary statistics and distribution of sales.
3. Compute Pearson correlation matrix between all advertising channels and sales.
4. Generate scatter plots with regression trendlines for each ad medium.
5. Examine overall budget allocation and pairwise relationships.

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

project_root = Path("..").resolve()
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from src.load_data import load_sales_dataframe

sns.set_theme(style="whitegrid", palette="deep")
plt.rcParams["figure.titlesize"] = 14
```

## 1. Load Data and Check Completeness

```python
df = load_sales_dataframe(file_path="../data/advertising.csv")
print("Shape:", df.shape)
df.info()
print("\nMissing values:\n", df.isnull().sum())
df.head()
```

## 2. Summary Statistics

```python
df[["tv", "radio", "newspaper", "total_spend", "sales"]].describe().T.round(2)
```

## 3. Target Distribution: Sales

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(df["sales"], kde=True, color="#1f77b4", ax=axes[0], bins=20)
axes[0].set_title("Sales Distribution (KDE + Histogram)")

sns.boxplot(x=df["sales"], color="#aec7e8", ax=axes[1])
axes[1].set_title("Sales Boxplot (Spread & Outliers)")
plt.tight_layout()
plt.show()
```

## 4. Correlation Heatmap

```python
plt.figure(figsize=(7, 5))
sns.heatmap(df[["tv", "radio", "newspaper", "total_spend", "sales"]].corr(), annot=True, fmt=".3f", cmap="coolwarm", vmin=-1, vmax=1, linewidths=1)
plt.title("Pearson Correlation Matrix")
plt.tight_layout()
plt.show()
```

## 5. Sales vs. Advertising Channels (Scatter + Regression Trendlines)

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
channels = ["tv", "radio", "newspaper"]
colors = ["#2980b9", "#27ae60", "#e67e22"]

for i, ch in enumerate(channels):
    sns.regplot(data=df, x=ch, y="sales", color=colors[i], line_kws={"color": "red"}, ax=axes[i])
    axes[i].set_title(f"Sales vs. {ch.upper()} Spend (r = {df[ch].corr(df['sales']):.3f})")
    axes[i].set_xlabel(f"{ch.upper()} Budget ($k)")
    axes[i].set_ylabel("Sales (k Units)")

plt.tight_layout()
plt.show()
```

## 6. Budget Breakdown by Channel

```python
channel_totals = [df["tv"].sum(), df["radio"].sum(), df["newspaper"].sum()]
plt.figure(figsize=(6, 6))
plt.pie(channel_totals, labels=["TV", "Radio", "Newspaper"], autopct="%1.1f%%", startangle=140, colors=["#3498db", "#2ecc71", "#e67e22"])
plt.title("Overall Budget Distribution by Marketing Channel")
plt.show()
```
